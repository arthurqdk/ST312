```python
import matplotlib.pyplot as plt
import numpy as np

# Data
visdata = {
    'BTC':3.1746,
    'BTC+EMA':3.1076,
    'BTC+Nasdaq':3.4182,
    'BTC+Google trends':3.1871,
    'BTC+EMA+Nasdaq':3.1913,
    'BTC+EMA+Google trends':3.5129,
    'BTC+Nasdaq+Google trends':3.1214,
    'BTC+EMA+Nasdaq+Google trends':3.1678,
    'BTC+EMA+Nasdaq+Google trends+latest_size':3.7868
    
}

# Extract data
labels = list(visdata.keys())
values = list(visdata.values())

# Set different colors for each bar
colors = plt.cm.get_cmap('tab10', len(visdata))(np.arange(len(visdata)))

# Create figure
plt.figure(figsize=(16, 10))

# Draw bar chart
bars = plt.bar(labels, values, color=colors, label='MAPE values')

# Add title and labels
plt.title('BiLSTM, Comparison of MAPE with window size=3', fontsize=18, fontweight='bold')
plt.xlabel('Input Features Combination', fontsize=14, fontweight='bold')
plt.ylabel('MAPE(%)', fontsize=14, fontweight='bold')

# Annotate each bar with its value, keeping four decimal places
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, yval, f'{yval:.4f}', ha='center', va='bottom', fontsize=12, fontweight='bold')

# Beautify the chart
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)



# Adjust layout to avoid label overlap
plt.tight_layout()

# Show figure
plt.show()
```
