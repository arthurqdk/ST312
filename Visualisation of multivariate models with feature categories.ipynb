# Bilstm


```python
import matplotlib.pyplot as plt
import numpy as np

# Data
visdata = {
    'BTC':3.1746,
    'Attention':3.8060,
    'Bitcoinall':6.9721,
    'Macro':3.8733,
    'On-chain':3.3347,
    'Technical':3.2569
}

# Extract labels and values
labels = list(visdata.keys())
values = list(visdata.values())

# Assign different colors to each bar
colors = plt.cm.get_cmap('tab10', len(visdata))(np.arange(len(visdata)))

# Create the figure
plt.figure(figsize=(8, 6))

# Plot the bar chart
bars = plt.bar(labels, values, color=colors, label='MAPE values')

# Add title and axis labels
plt.title('BiLSTM,  comparison of MAPE, window size = 3', fontsize=18, fontweight='bold')
plt.xlabel('feature category', fontsize=14, fontweight='bold')
plt.ylabel('MAPE(%)', fontsize=14, fontweight='bold')

# Annotate values on top of bars (4 decimal places)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, yval, f'{yval:.4f}', ha='center', va='bottom', fontsize=12, fontweight='bold')

# Beautify axes
plt.xticks(rotation=45, ha='right', fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Adjust layout to prevent label overlap
plt.tight_layout()

# Show plot
plt.show()
```
